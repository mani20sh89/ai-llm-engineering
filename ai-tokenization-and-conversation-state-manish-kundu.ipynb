# Tokenization and Conversation State

This notebook demonstrates how text is tokenized and why chat models do not retain conversation history between independent API requests.

## 1. Tokenization

```python
import tiktoken

encoding = tiktoken.get_encoding("o200k_base")
text = "Hi, my name is Ed and I like banoffee pie."
tokens = encoding.encode(text)

print("Token IDs:", tokens)
print("Token count:", len(tokens))
```

```python
for token_id in tokens:
    print(f"{token_id}: {encoding.decode([token_id])!r}")
```

## 2. Chat API requests

Each API request is independent. To preserve context, include the relevant conversation history in the next request.

Set `OPENAI_API_KEY` in your environment before running these cells.

```python
from openai import OpenAI

client = OpenAI()
```

### First request

```python
messages = [
    {"role": "system", "content": "You are a helpful assistant."},
    {"role": "user", "content": "Hi! My name is Ed."},
]

response = client.chat.completions.create(
    model="gpt-4.1-mini",
    messages=messages,
)

print(response.choices[0].message.content)
```

### Follow-up request without history

```python
messages = [
    {"role": "system", "content": "You are a helpful assistant."},
    {"role": "user", "content": "What is my name?"},
]

response = client.chat.completions.create(
    model="gpt-4.1-mini",
    messages=messages,
)

print(response.choices[0].message.content)
```

### Follow-up request with history

Include the earlier messages so the model has the context needed to answer.

```python
messages = [
    {"role": "system", "content": "You are a helpful assistant."},
    {"role": "user", "content": "Hi! My name is Ed."},
    {"role": "assistant", "content": "Hi Ed! How can I help?"},
    {"role": "user", "content": "What is my name?"},
]

response = client.chat.completions.create(
    model="gpt-4.1-mini",
    messages=messages,
)

print(response.choices[0].message.content)
```

## Key takeaways

- Tokenizers convert text into token IDs.
- A chat model does not automatically receive previous requests as context.
- Include conversation history in each request when the model needs it.
- More conversation history means more input tokens are processed.
